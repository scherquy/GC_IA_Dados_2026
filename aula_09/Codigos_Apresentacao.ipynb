```python
arquivo = open("cadastro_mini.csv", "r", encoding="UTF-8")
planilha = arquivo.readlines()
arquivo.close()

for linha in planilha:
    print(linha)
```

```python
for linha in planilha:
    linha = linha.strip()
    linha_separada = linha.split(";")

    status = linha_separada[0].strip().lower().replace("ativoo", "ativo").replace("inativoo", "inativo").title()
    nome = linha_separada[1].strip().replace("  ", " ").title()
    regiao = linha_separada[2].strip().title()

    print(f"{status:<7} | {nome:<16} | {regiao:<6}")
```
